```python
#Question 1
def linear_merge(list1, list2):
    merged_list = []
    i, j = 0, 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1
    merged_list.extend(list1[i:])
    merged_list.extend(list2[j:])
    return merged_list

# Example usage:
list1 = [1, 3, 5]
list2 = [2, 4, 6]
print(linear_merge(list1, list2))
```

```python
#Question 5
def divide(P, Q):
    try:
        result = P / Q
    except ZeroDivisionError:
        return "Division by zero is not allowed"
    except TypeError:
        return "Both P and Q must be numbers"
    return result

#Test Cases
print(divide(10, 2))
print(divide(10, 0))
print(divide(10, 'a'))
```
