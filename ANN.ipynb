```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import fashion_mnist

from sklearn.metrics import confusion_matrix, classification_report
```

```python
(X_train, Y_train), (X_test, Y_test) = fashion_mnist.load_data()
```

```python
Y_train
```

```python
# specify the number of rows and columns you want to see
num_rows, num_cols = 5, 5
fig, axes = plt.subplots(num_rows, num_cols, figsize=(10, 10))
fig.suptitle("Sample Images from Fashion MNIST Dataset", fontsize=16)


# get a segment of the dataset
X_train_segment = X_train[:10000]
Y_train_segment = Y_train[:10000]
X_test_segment = X_test[:2000]
Y_test_segment = Y_test[:2000]
print("Training data segment shape:", X_train_segment.shape)
print("Test data segment shape:", X_test_segment.shape)

# plot images

for i, ax in enumerate(axes.flat):
    ax.imshow(X_train[i], cmap='gray')
    ax.axis('off')
plt.show()
```

```python
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train[1]
```

```python
 from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.datasets import fashion_mnist
from sklearn.metrics import confusion_matrix, classification_report

model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')  # 10 classes for Fashion MNIST
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

model.fit(X_train, Y_train, epochs=5, validation_data=(X_test, Y_test))

predictions = model.predict(X_test)
y_pred = np.argmax(predictions, axis=1)  # Convert model output to label predictions

# Calculate confusion matrix
cm = confusion_matrix(Y_test, y_pred)
```

```python
np.unique(Y_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report

# Draw confusion matrix
plt.figure(figsize=(8, 6))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.colorbar()
plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.show()

```

```python
# Calculate and print the accuracy
accuracy = np.trace(cm) / float(np.sum(cm))
print("Accuracy of the model: {:.2f}%".format(accuracy * 100))
```

```python

# Generate classification report
print("\nClassification Report:")
print(classification_report(Y_test, y_pred))
```
