```python
import copy
import math

X = "X"
O = "O"
EMPTY = None

def initial_state():
    """Returns the starting state of the board."""
    return [[EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY]]

def player(board):
    """Returns the player who has the next turn on a board."""
    count = sum(1 for row in board for cell in row if cell)
    return O if count % 2 else X

def actions(board):
    """Returns set of all possible actions (i, j) available on the board."""
    return {(i, j) for i in range(3) for j in range(3) if board[i][j] == EMPTY}

def result(board, action):
    """Returns the board that results from making move (i, j) on the board."""
    curr_player = player(board)
    result_board = copy.deepcopy(board)
    (i, j) = action
    result_board[i][j] = curr_player
    return result_board

def winner(board):
    """Returns the winner of the game, if there is one."""
    for row in board:
        if row[0] and all(cell == row[0] for cell in row):
            return row[0]
    for col in range(3):
        if board[0][col] and all(board[row][col] == board[0][col] for row in range(3)):
            return board[0][col]
    if board[0][0] and all(board[i][i] == board[0][0] for i in range(3)):
        return board[0][0]
    if board[0][2] and all(board[i][2 - i] == board[0][2] for i in range(3)):
        return board[0][2]
    return None

def terminal(board):
    """Returns True if game is over, False otherwise."""
    return winner(board) is not None or all(cell is not EMPTY for row in board for cell in row)

def utility(board):
    """Returns 1 if X has won the game, -1 if O has won, 0 otherwise."""
    win = winner(board)
    if win == X:
        return 1
    elif win == O:
        return -1
    return 0

def alpha_beta_pruning(board, alpha=-math.inf, beta=math.inf):
    """Returns the optimal action for the current player using alpha-beta pruning."""
    if terminal(board):
        return None, utility(board)

    best_action = None

    if player(board) == X:  # Maximize for X
        max_eval = -math.inf
        for action in actions(board):
            eval = alpha_beta_pruning(result(board, action), alpha, beta)[1]
            if eval > max_eval:
                max_eval = eval
                best_action = action
            alpha = max(alpha, eval)
            if beta <= alpha:
                break
        return best_action, max_eval
    else:  # Minimize for O
        min_eval = math.inf
        for action in actions(board):
            eval = alpha_beta_pruning(result(board, action), alpha, beta)[1]
            if eval < min_eval:
                min_eval = eval
                best_action = action
            beta = min(beta, eval)
            if beta <= alpha:
                break
        return best_action, min_eval

if __name__ == "__main__":
    user = input("Choose a player (X/O): ").strip().upper()
    board = initial_state()

    while True:
        print("Current board:")
        for row in board:
            print(row)

        if terminal(board):
            win = winner(board)
            if win is None:
                print("Game Over: Tie.")
            else:
                print(f"Game Over: {win} wins.")
            break

        playr = player(board)

        if user != playr:  # AI turn
            move, _ = alpha_beta_pruning(board)
            if move is not None:
                board = result(board, move)
                print(f"AI plays: {move}")
        else:  # User turn
            print("Your turn. Enter the position to move (row, col):")
            while True:
                try:
                    i = int(input("Row (0-2): "))
                    j = int(input("Col (0-2): "))
                    if (i, j) in actions(board):
                        board = result(board, (i, j))
                        break
                    else:
                        print("Invalid move. Try again.")
                except (ValueError, IndexError):
                    print("Invalid input. Please enter numbers between 0 and 2.")

        # Check for terminal state again after player's turn
        if terminal(board):
            print("Current board:")
            for row in board:
                print(row)
            win = winner(board)
            if win is None:
                print("Game Over: Tie.")
            else:
                print(f"Game Over: {win} wins.")
            break
```
