```python
#Question 1
def chickens(count):
    if count >= 10:
        return "Number of chickens: many"
    else:
        return f"Number of chickens: {count}"

# Examples
print(chickens(5))
print(chickens(23))
```

```python
#Question 2
def both_ends(s):
    if len(s) < 3:
        return ""
    return s[:3] + s[-3:]

# Examples
print(both_ends("intelligence"))
print(both_ends("hi"))
```

```python
#Question 3
def replace_first_char(s):
    first_char = s[0]
    return first_char + s[1:].replace(first_char, '@')

# Examples
print(replace_first_char("Ooogle"))
print(replace_first_char("banana"))
```

```python
#Question 4
def string_jumble(a, b):
    new_a = b[:2] + a[2:]
    new_b = a[:2] + b[2:]
    return new_a + " " + new_b

# Examples
print(string_jumble("mix", "pod"))
print(string_jumble("dog", "dinner"))
```

```python
#question 5
def match_ends(words):
    count = 0
    for word in words:
        if len(word) >= 2 and word[0] == word[-1]:
            count += 1
    return count

# Examples
print(match_ends(["aba", "xyz", "aa", "x", "bbb"]))
print(match_ends(["", "a", "ab", "abc", "xyx"]))
```

```python
#Question 6
def sort_last(tuples):
    return sorted(tuples, key=lambda x: x[-1])

# Examples
print(sort_last([(1, 7), (1, 3), (3, 4, 5), (2, 2)]))
print(sort_last([(2, 5), (1, 2), (4, 4), (2, 3)]))
```
