```python
from collections import deque
def loadFile():
  with open('grid.txt') as file:
    matrix = [[str(digit) for digit in line.split()] for line in file]
  return matrix

def BFS(matrix):
    rows = len(matrix)
    cols = len(matrix[0])

    ri, cj = None, None
    goals = []


    for i in range(rows):
        for j in range(cols):
            if matrix[i][j] == 'S':
                ri, cj = i, j
            elif matrix[i][j] == 'G':
                goals.append((i, j))



    queue = deque([(ri, cj, [])])

    visited = set()

    while queue:
        curr_i, curr_j, path = queue.popleft()

        if (curr_i, curr_j) in visited:
            continue

        visited.add((curr_i, curr_j))

        for new_i, new_j in [(curr_i+1, curr_j), (curr_i-1, curr_j), (curr_i, curr_j+1), (curr_i, curr_j-1)]:
            if 0 <= new_i < rows and 0 <= new_j < cols and matrix[new_i][new_j] != '1' and (new_i, new_j) not in visited:
                new_path = path + [(curr_i, curr_j)]
                if matrix[new_i][new_j] == 'G':
                    return new_path + [(new_i, new_j)]
                queue.append((new_i, new_j, new_path))


    return -1

def DFS(matrix):
    rows = len(matrix)
    cols = len(matrix[0])

    def dfs(curr_i, curr_j, path, visited):
        if (curr_i, curr_j) in visited:
            return None

        visited.add((curr_i, curr_j))
        path.append((curr_i, curr_j))

        if matrix[curr_i][curr_j] == 'G':
            return path

        for new_i, new_j in [(curr_i+1, curr_j), (curr_i-1, curr_j), (curr_i, curr_j+1), (curr_i, curr_j-1)]:
            if 0 <= new_i < rows and 0 <= new_j < cols and matrix[new_i][new_j] != '1':
                result = dfs(new_i, new_j, path.copy(), visited.copy())
                if result:
                    return result

        return None

    ri, cj = None, None
    for i in range(rows):
        for j in range(cols):
            if matrix[i][j] == 'S':
                ri, cj = i, j
                break
        if ri is not None and cj is not None:
            break

    if ri is None or cj is None:

        return -1

    return dfs(ri, cj, [], set())




matrix = loadFile()
BFSpath = BFS(matrix)
DFSpath= DFS(matrix)


print("BFS",BFSpath)
print("DFS",DFSpath)
```

```python
def loadInputData():
  with open('cube.txt') as file:
    matrix = [[str(digit) for digit in line.split()] for line in file]
  return matrix




cube = loadInputData()
start = goal = None
for i, row in enumerate(cube):
    for j, cell in enumerate(row):
        if cell == 'S':
            start = (i, j)
        elif cell == 'G':
            goal = (i, j)



def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def get_neighbors(pos, cube):
    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    neighbors = []
    for d in directions:
        nr, nc = pos[0] + d[0], pos[1] + d[1]
        if 0 <= nr < len(cube) and 0 <= nc < len(cube[0]) and cube[nr][nc] != '1':
            neighbors.append((nr, nc))
    return neighbors

def a_star_search(start, goal, cube):
    from queue import PriorityQueue
    frontier = PriorityQueue()
    frontier.put((0, start))
    came_from = {start: None}
    cost_so_far = {start: 0}

    while not frontier.empty():
        current = frontier.get()[1]

        if current == goal:
            break

        for next in get_neighbors(current, cube):
            new_cost = cost_so_far[current] + (1 if cube[next[0]][next[1]] != '2' else 0)  # Jumping over short walls costs 0
            if next not in cost_so_far or new_cost < cost_so_far[next]:
                cost_so_far[next] = new_cost
                priority = new_cost + heuristic(goal, next)
                frontier.put((priority, next))
                came_from[next] = current

    path = []
    current = goal
    while current != start:
        path.append(current)
        current = came_from[current]
    path.append(start)
    path.reverse()
    return path



start = goal = None
for i, row in enumerate(cube):
    for j, cell in enumerate(row):
        if cell == 'S':
            start = (i, j)
        elif cell == 'G':
            goal = (i, j)

path = a_star_search(start, goal, cube)
print("The path after A* search is:")
print(path)
```
